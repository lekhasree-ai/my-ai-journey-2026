Day 21: Feature engineering, filtering, and sorting in Pandas

```python
import pandas as pd

data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Kabir"],
    "Math": [88, 95, 76, 91, 84],
    "Physics": [84, 92, 79, 89, 81],
    "CS": [92, 98, 85, 94, 89]
}

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

# 1. Feature Engineering: Create Total and Average columns
df["Total"] = df["Math"] + df["Physics"] + df["CS"]
df["Average"] = (df["Total"] / 3).round(2)
print("The updated DataFrame:\n", df)

# 2. Filtering: Students who scored > 90 in CS
top_cs = df[df["CS"] > 90]
print("\nThe top students in CS:\n", top_cs)

# 3. Sorting: Rank all students from highest to lowest Total
leaderboard = df.sort_values(by="Total", ascending=False)
print("\nRankings of all students:\n", leaderboard)
```
