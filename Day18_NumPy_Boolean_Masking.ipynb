```python
import numpy as np

scores = np.array([62, 88, 95, 74, 81, 59, 91, 79])

# 1. Create and print the True/False Boolean mask
high_mask = scores >= 80
print("Boolean Mask:", high_mask)

# 2. Filter out and save only the scores 80 or higher
high_scores = scores[scores >= 80]
print("Scores 80 or higher:", high_scores)

# 3. Filter passing scores (>= 65) and calculate their average
passing_scores = scores[scores >= 65]
print("Passing scores:", passing_scores)
print("Average of passing scores:", np.mean(passing_scores))
```
