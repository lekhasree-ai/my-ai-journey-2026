```python
def get_top_student(students_list):
    top_name = students_list[0]["name"]
    highest_score = students_list[0]["score"]

    for student in students_list:
        if student["score"] > highest_score:
            highest_score = student["score"]
            top_name = student["name"]

    return top_name

class_data = [
    {"name": "Alex", "id": 101, "score": 85},
    {"name": "Sam", "id": 102, "score": 96},
    {"name": "Jordan", "id": 103, "score": 78}
]

best_student = get_top_student(class_data)
print("The top student is:", best_student)
```
