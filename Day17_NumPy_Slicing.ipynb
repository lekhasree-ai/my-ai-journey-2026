```python
import numpy as np
#a 4x3 scores matrix (4 students as rows, 3 subjects as columns: Math, Physics, CS)
scores = np.array([
    [85, 90, 88],   # Row 0
    [78, 82, 80],   # Row 1
    [92, 95, 94],   # Row 2
    [74, 67, 85]    # Row 3
])
print("Student 2's Physics score:",scores[1,1])
print("All marks for Student 3:",scores[2,:])
print("All students' CS marks:",scores[:,2])
print("The average CS score:",np.mean(scores[:,2]))
```
