```python
import pandas as pd

data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Kabir", "Ananya"],
    "Section": ["A", "B", "A", "B", "A", "B"],
    "Math": [88, 95, 76, 91, 84, 89],
    "Physics": [84, 92, 79, 89, 81, 87],
    "CS": [92, 98, 85, 94, 89, 96]
}

df = pd.DataFrame(data)

# 1. Engineer Total column
df["Total"] = df["Math"] + df["Physics"] + df["CS"]
print("The complete DataFrame:\n", df)

# 2. Average marks per section (double brackets for multiple columns!)
section = df.groupby("Section")[["Math", "Physics", "CS", "Total"]].mean().round(2)
print("\nThe average by section and subjects:\n", section)

# 3. Highest Total score in each section
high_total = df.groupby("Section")["Total"].max()
print("\nThe highest Total score per section:\n", high_total)

# 4. Student count per section
no_students = df.groupby("Section")["Name"].count()
print("\nThe total number of students per section:\n", no_students)
```
