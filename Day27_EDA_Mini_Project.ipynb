```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Kabir", "Ananya", "Vikram", "Meera"],
    "Section": ["A", "A", "A", "A", "B", "B", "B", "B"],
    "Study_Hours": [2.0, 5.5, np.nan, 6.5, 3.0, 7.0, 4.5, np.nan],
    "Attendance": [72, 95, 78, 96, 80, 98, 85, 90],
    "Exam_Score": [61, 88, 70, 93, 68, 96, 79, 84]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# Stage 2: Clean missing data
mean_study = df["Study_Hours"].mean().round(1)
df["Study_Hours"] = df["Study_Hours"].fillna(mean_study)
print("\nCleaned DataFrame (Missing values filled):\n", df)
print("Remaining missing values check:", df.isnull().sum().sum())

# Stage 3: Aggregate insights using .groupby()
section_summary = df.groupby("Section")[["Study_Hours", "Attendance", "Exam_Score"]].mean().round(2)
print("\nAverage metrics by Section:\n", section_summary)

# Stage 4: Visualizations
# Chart 1: Scatter plot with regression trendline (Quotes added to title!)
sns.regplot(data=df, x="Study_Hours", y="Exam_Score", color="teal")
plt.title("Scatter Plot with Regression Trendline")
plt.show()

# Chart 2: Correlation heatmap (Quotes added to title!)
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
