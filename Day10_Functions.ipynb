```python
def check_grade(score):
    if score >= 50:
        print("Score:", score, "-> Passed! 👍")
    else:
        print("Score:", score, "-> Needs improvement! 💪")

# Testing the function with different scores:
check_grade(85)
check_grade(42)
check_grade(90)
```
