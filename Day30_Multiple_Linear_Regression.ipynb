```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Features: [Study_Hours, Attendance]
X = [
    [2.0, 72],
    [5.5, 95],
    [3.0, 78],
    [6.5, 96],
    [4.0, 85],
    [7.0, 98],
    [4.5, 82],
    [6.0, 90]
]

# Target: Exam_Score
y = [61, 88, 70, 93, 82, 97, 79, 91]
model = LinearRegression()
model.fit(X, y)

print("Learned Coefficients (Slopes for Study Hours & Attendance):", model.coef_)
print("Learned Intercept (c):", model.intercept_)
# Predict score for a student with 5.0 study hours and 88% attendance
new_student = [[5.0, 88]]
prediction = model.predict(new_student)
print(f"Predicted exam score for 5.0 hours study & 88% attendance: {prediction[0]:.2f}")
```
