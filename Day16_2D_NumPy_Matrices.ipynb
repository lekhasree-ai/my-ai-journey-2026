```python
import numpy as np
scores = np.array([
    [85, 90, 88],   # Student 1
    [78, 82, 80],   # Student 2
    [92, 95, 94],   # Student 3
    [74, 67, 85]    # Student 4
])
print("The dimensions of matrix:",(scores.shape))
print("Each student's average:",np.mean(scores, axis=1))
print("Each subject's average:",np.mean(scores, axis=0))
print("The overall highest score in the table:",np.max(scores))
```
