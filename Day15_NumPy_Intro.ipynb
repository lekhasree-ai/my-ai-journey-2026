```python
import numpy as np

# Regular Python list
marks_list = [85, 92, 78, 90, 88]

# Convert to a NumPy array
marks_array = np.array(marks_list)

# Built-in AI math functions (no loops required!)
print("Average Mark:", np.mean(marks_array))
print("Highest Mark:", np.max(marks_array))

# Vectorized math: add 5 bonus marks to everyone at once
boosted_marks = marks_array + 5
print("Boosted Marks:", boosted_marks)
```
