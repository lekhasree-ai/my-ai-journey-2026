```python
def calculate_average(marks_list):
    total = sum(marks_list)
    avg = total / len(marks_list)
    return avg  # Sends the calculated average back out

my_scores = [85, 90, 78, 92, 88]
final_avg = calculate_average(my_scores)

print("The calculated average:", final_avg)
```
