Day 22: Detecting and cleaning missing values (NaN) in Pandas

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Kabir"],
    "Math": [88, 95, 76, 91, 84],
    "Physics": [84, 92, np.nan, 89, 81],
    "CS": [92, 98, 85, 94, np.nan]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# 1. Detect missing values
mv = df.isnull().sum()
print("\nMissing values per column:\n", mv)

# 2. Method 1: Drop rows with any NaN
df_dropped = df.dropna()
print("\nDataFrame after dropping incomplete rows:\n", df_dropped)

# 3. Method 2: Calculate column averages and fill the NaN gaps
phy_mean = df["Physics"].mean().round(1)
cs_mean = df["CS"].mean().round(1)

df["Physics"] = df["Physics"].fillna(phy_mean)
df["CS"] = df["CS"].fillna(cs_mean)

print("\nCleaned DataFrame (after mean imputation):\n", df)
print("\nRemaining missing values:\n", df.isnull().sum())
```
