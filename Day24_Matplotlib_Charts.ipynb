Day 24: Creating Bar Charts and Line Plots with Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Kabir"],
    "CS": [92, 98, 85, 94, 89]
}
df = pd.DataFrame(data)
plt.bar(df["Name"],df["CS"])
plt.title("Computer Science Marks by Student")
plt.xlabel("Student Name")
plt.ylabel("CS Score")
plt.show()
epochs = [1, 2, 3, 4, 5]
accuracy = [62, 74, 81, 88, 93]
plt.plot(epochs, accuracy, marker="o", color="green")
plt.title("AI Model Accuracy Over Training Epochs")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.show()
```
