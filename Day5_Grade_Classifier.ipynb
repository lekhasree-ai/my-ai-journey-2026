```python
score = int(input("Enter your total score: "))

if score >= 90:
    print("Grade A - Outstanding! 🌟")
elif score >= 75:
    print("Grade B - Distinction! 🎉")
elif score >= 50:
    print("Grade C - Passed! 👍")
else:
    print("Needs improvement, keep trying! 💪")
```
