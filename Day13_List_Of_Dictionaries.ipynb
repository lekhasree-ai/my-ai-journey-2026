```python
students = [
    {"name": "Alex", "id": 101, "score": 85},
    {"name": "Sam", "id": 102, "score": 92},
    {"name": "Jordan", "id": 103, "score": 78}
]

for student in students:
    print("Name:", student["name"], "| Score:", student["score"])
```
