```python
import numpy as np

# 1. Generate 15 random exam scores between 50 and 100
array_1d = np.random.randint(50, 101, size=15)
print("Flat one dimensional array:", array_1d)

# 2. Reshape into 5 students (rows) and 3 subjects (columns)
dataset = array_1d.reshape(5, 3)
print("The reshaped dataset:\n", dataset)

# 3. Inspect shape (attribute, no parentheses) and student averages
print("The shape of the dataset:", dataset.shape)
print("Each student's average across their 3 subjects:", np.mean(dataset, axis=1))
```
