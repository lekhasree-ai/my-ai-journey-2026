```python
import numpy as np
import matplotlib.pyplot as plt
study_hours = [1.5, 2.0, 3.0, 3.5, 4.5, 5.0, 6.0, 7.0, 8.0]
exam_scores = [55, 62, 68, 71, 77, 82, 86, 92, 96]
plt.scatter(study_hours, exam_scores, color="darkorange", s=80)
plt.title("Study Hours vs. Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
np.random.seed(42)
scores = np.random.normal(loc=78, scale=10, size=100)
plt.hist(scores, bins=8, color="skyblue", edgecolor="black")
plt.title("Distribution of 100 Student Exam Scores")
plt.xlabel("Score Range")
plt.ylabel("Number of Students")
plt.show()
```
