```python
import pandas as pd

data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Kabir"],
    "Math": [88, 95, 76, 91, 84],
    "Physics": [84, 92, 79, 89, 81],
    "CS": [92, 98, 85, 94, 89]
}

df = pd.DataFrame(data)
print("The DataFrame:\n", df)
print("\nThe first three rows:\n", df.head(3))
print("\nThe statistical summary of all numeric columns:\n", df.describe())

cs = df["CS"]
print("\nThe CS column:\n", cs)
print("\nThe mean of CS marks:", df["CS"].mean())
```
