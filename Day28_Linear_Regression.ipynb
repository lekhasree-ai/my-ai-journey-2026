```python
# Hours studied (Features)
X = [[1], [2], [3], [4], [5], [6], [7], [8]]

# Actual exam scores received (Target)
y = [52, 61, 70, 80, 89, 95, 98, 100]
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
# Hours studied (Features)
X = [[1], [2], [3], [4], [5], [6], [7], [8]]

# Actual exam scores received (Target)
y = [52, 61, 70, 80, 89, 95, 98, 100]
model = LinearRegression()
model.fit(X, y)

print("Learned Slope (m):", model.coef_[0])
print("Learned Intercept (c):", model.intercept_)
# Predict score for a student who studied for 5.5 hours
new_student = [[5.5]]
prediction = model.predict(new_student)
print(f"Predicted score for 5.5 study hours: {prediction[0]:.2f}")
plt.scatter(X, y, color="purple", label="Actual Data")
plt.plot(X, model.predict(X), color="orange", linewidth=2, label="AI Trendline")
plt.title("Simple Linear Regression: Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
